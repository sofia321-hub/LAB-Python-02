```python
numeros = [2,5,10,15,20]
palabras = ["hola","universidad","python","programacion","sol"]

tuplas = [(n, n**2) for n in numeros]
print(tuplas)

cubos = [n**3 for n in numeros if n > 10]
print(cubos)

pares = [n for n in numeros if n % 2 == 0]
print(pares)

valores = [f"valor: {n}" for n in numeros]
print(valores)

fahrenheit = [(c*9/5)+32 for c in numeros]
print(fahrenheit)

mas5 = [p for p in palabras if len(p) > 5]
print(mas5)

iniciales = [p[0] for p in palabras]
print(iniciales)

par_impar = [
    "par" if n % 2 == 0 else "impar"
    for n in numeros
]

print(par_impar)

matriz = [[1,2,3],[4,5,6],[7,8,9]]

plana = [
    x
    for fila in matriz
    for x in fila
]

print(plana)
```
