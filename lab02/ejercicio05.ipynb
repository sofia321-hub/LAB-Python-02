```python
from functools import reduce

nums=[2,3,4]

resultado=reduce(
lambda a,b:a*b,
nums
)

print(resultado)


palabras=["hola","mundo"]

texto=reduce(
lambda a,b:a+b,
palabras
)

print(texto)


def mayor(a,b):
    return a if a>b else b


nums=[10,2,50,8]

print(
reduce(
mayor,
nums
)
)
```
