```python
from math import pi

celsius=[10,20,30]

f=list(
map(lambda x:(x*9/5)+32,celsius)
)

print(f)


l1=[1,2,3]
l2=[4,5,6]

suma=list(
map(lambda a,b:a+b,l1,l2)
)

print(suma)


radios=[1,2,3]

areas=list(
map(lambda r:pi*r*r,radios)
)

print(areas)


productos={
"pan":1000,
"leche":2500,
"cafe":5000
}

nuevo=list(
map(
lambda x:(x[0],x[1]*1.10),
productos.items()
)
)

print(nuevo)


matriz=[[1,2],[3,4]]

resultado=list(
map(
lambda fila:
list(
map(lambda x:x*10,fila)
),
matriz
)
)

print(resultado)
```
