```python
multiplo3 = lambda x: x % 3 == 0
print(multiplo3(9))
print(multiplo3(10))

cubo = lambda x: x**3
print(cubo(4))

producto = lambda a, b: a * b
print(producto(5,8))

mayor = lambda a, b: a if a > b else b
print(mayor(10,7))

empieza_a = lambda palabra: palabra.lower().startswith("a")
print(empieza_a("Avión"))
print(empieza_a("casa"))

fahrenheit = lambda c: (c*9/5)+32
print(fahrenheit(30))

funciones = [
    lambda x: x*2,
    lambda x: x+10,
    lambda x: x**2
]

numero = int(input("Ingrese un número: "))

for f in funciones:
    print(f(numero))
```
