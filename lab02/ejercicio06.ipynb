```python
palabras=["zorro","ana","casa"]

print(
sorted(palabras)
)


tuplas=[
(5,"a"),
(1,"b"),
(3,"c")
]

print(
sorted(
tuplas,
key=lambda x:x[0]
)
)


productos=[
("Camisa",30),
("Pantalon",55),
("Medias",10),
("Chaqueta",80)
]

print(
sorted(
productos,
key=lambda x:(-x[1],x[0])
)
)


print(
sorted(
palabras,
key=len
)
)


nums=[5,2,8,1,7]

print(
sorted(
nums,
key=lambda x:x%2
)
)


texto="banana"

frecuencia={}

for letra in texto:
    frecuencia[letra]=frecuencia.get(letra,0)+1

print(
sorted(
frecuencia.items(),
key=lambda x:x[1]
)
)


empleados=[
("Maria","Ventas",30),
("Luis","Tecnologia",25),
("Ana","Ventas",25),
("Pedro","Tecnologia",28)
]

print(
sorted(
empleados,
key=lambda x:(x[1],x[2])
)
)
```
